```python

# 1. IMPORTS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```


# 2. DATA COLLECTION & LOADING


```python

housing = fetch_california_housing(as_frame=True)

df = housing.frame

print("First 5 Rows")
print(df.head())
```


# 3. QUICK CHECK OF DATA


```python

print("\nDataset Information")
print(df.info())

print("\nStatistical Summary")
print(df.describe())

print("\nColumn Names")
print(df.columns)
```

4. EDA & DATA PREPROCESSING

```python

# Check missing values
print("\nMissing Values")
print(df.isnull().sum())

# Correlation Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# Target Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['MedHouseVal'], kde=True)
plt.title("Distribution of Median House Value")
plt.xlabel("MedHouseVal")
plt.show()

# Boxplot
plt.figure(figsize=(8,3))
sns.boxplot(x=df['MedHouseVal'])
plt.title("Median House Value Boxplot")
plt.show()
```

```python

# EDA INSIGHTS


print("""
EDA INSIGHTS

• No missing values were found in the dataset.
• Most house values are concentrated in lower price ranges.
• The distribution is right-skewed.
• Some high-value outliers exist.
• Certain features have stronger correlations with
  house values than others.
""")
```

```python

# SCATTERPLOT FUNCTION

def scatter_plot(feature):

    plt.figure(figsize=(6,4))

    plt.scatter(
        df[feature],
        df['MedHouseVal'],
        alpha=0.4
    )

    plt.xlabel(feature)
    plt.ylabel("MedHouseVal")
    plt.title(f"{feature} vs MedHouseVal")

    plt.show()


# Required Features

features = [
    'MedInc',
    'AveRooms',
    'AveOccup',
    'HouseAge'
]

for feature in features:
    scatter_plot(feature)


print("""
Scatter Plot Interpretation

• MedInc shows a strong positive relationship with
  house values.

• AveRooms shows a moderate positive relationship.

• AveOccup has a weaker relationship.

• HouseAge shows some positive influence but is
  less important than MedInc.

Median income appears to be one of the strongest
predictors of house value.
""")
```

5. ML MODEL TRAINING

```python

X = df.drop('MedHouseVal', axis=1)

y = df['MedHouseVal']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Linear Regression Model

model = LinearRegression()

model.fit(X_train, y_train)

print("\nModel Training Complete")
```


# 6. MODEL EVALUATION


```python

y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("\nModel Evaluation")

print("RMSE =", rmse)

print("R² Score =", r2)


print("""
Evaluation Interpretation

• RMSE measures prediction error.
  Lower values indicate better performance.

• R² measures how much variation in house values
  is explained by the model.

• A higher R² means the model fits the data well.
""")
```


# 7. FEATURE IMPORTANCE


```python

coefficients = pd.Series(
    model.coef_,
    index=X.columns
)

coefficients.sort_values().plot(
    kind='barh',
    figsize=(8,6)
)

plt.title("Linear Regression Coefficients")
plt.show()


print("""
Feature Importance Interpretation

• Features with larger coefficients have greater
  influence on house prices.

• MedInc is typically one of the strongest positive
  predictors.

• Location-related variables such as Latitude and
  Longitude also influence prices significantly.

• Income and geographic location are major drivers
  of California housing prices.
""")
```

8. MODEL PREDICTION

```python

new_data = pd.DataFrame([{

    'MedInc': 3,
    'HouseAge': 30,
    'AveRooms': 6,
    'AveBedrms': 1,
    'Population': 1500,
    'AveOccup': 3,
    'Latitude': 34,
    'Longitude': -118

}])

predicted_value = model.predict(new_data)

print("\nPredicted Median House Value:")
print(predicted_value[0])


print("""
Prediction Interpretation

The model predicts the median house value
for the provided housing characteristics.

This prediction is based on patterns learned
from the California Housing dataset.
""")
```
